### PyGenStabilityOne Example Code

```python
from pygenstability_one import pygenstability_one
import networkx as nx
```

```python
graph = nx.read_gml("ABCD_0.gml", destringizer=int)

coms = pygenstability_one(graph)
len(coms)
```

```python
G = nx.barbell_graph(6,0)
H = nx.disjoint_union(G,G)
```

```python
import matplotlib.pyplot as plt
plt.figure(figsize=(10,8))
# nx.draw_circular(H,with_labels=True)
nx.draw_networkx(H, pos=nx.circular_layout(H), with_labels=True)
```

```python
coms = pygenstability_one(H)
len(coms)
```
