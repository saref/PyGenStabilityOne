### PyGenStabilityOne Example Code

Import Dependencies

```python
from pygenstability_one import pygenstability_one
import matplotlib.pyplot as plt
import networkx as nx
```

Example using a graph read from file

```python
graph = nx.read_gml("ABCD_0.gml", destringizer=int)

plt.figure(figsize=(10,8))
nx.draw_networkx(graph, pos=nx.spring_layout(graph), with_labels=True)
```

```python
coms = pygenstability_one(graph)
len(coms)
```

Example using a graph constructed with networkx

```python
G = nx.barbell_graph(6,0)
H = nx.disjoint_union(G,G)

plt.figure(figsize=(10,8))
nx.draw_networkx(H, pos=nx.circular_layout(H), with_labels=True)
```

```python
coms = pygenstability_one(H)
print(coms)
```
